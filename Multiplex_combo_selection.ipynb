```python
df = pd.DataFrame(data, columns=[
    "Name",
    "Forward Primer",
    "Reverse Primer",
    "Probe",
    "Forward Tm",
    "Reverse Tm",
    "Probe Tm",
    "Product Size"
])

display(df)
```

```python
# ==========================================================
# STEP 2: SETTINGS
# ==========================================================

PLEX = 3          # choose 3 or 4
TOP_RESULTS = 20
REACTION_TEMP = 61
```

```python
# ==========================================================
# STEP 3: FUNCTIONS
# ==========================================================

def calc_dg(seq1, seq2):
    try:
        result = primer3.calc_heterodimer(
            seq1,
            seq2,
            temp_c=REACTION_TEMP,
            mv_conc=50,
            dv_conc=1.5,
            dna_conc=250
        )
        return abs(result.dg) / 1000
    except:
        return 0

def score_combo(combo):

    primers = []
    probes = []

    for x in combo:
        primers.append(x["Forward Primer"])
        primers.append(x["Reverse Primer"])
        probes.append(x["Probe"])

    # -----------------------------------
    # Primer-Primer interactions
    # -----------------------------------
    pp = []
    for a,b in itertools.combinations(primers,2):
        pp.append(calc_dg(a,b))

    # -----------------------------------
    # Primer-Probe interactions
    # -----------------------------------
    pq = []
    for p in primers:
        for q in probes:
            pq.append(calc_dg(p,q))

    # -----------------------------------
    # Probe-Probe interactions
    # -----------------------------------
    qq = []
    for a,b in itertools.combinations(probes,2):
        qq.append(calc_dg(a,b))

    # -----------------------------------
    # Tm spread
    # -----------------------------------
    tms = []
    sizes = []

    for x in combo:
        tms.extend([
            x["Forward Tm"],
            x["Reverse Tm"],
            x["Probe Tm"]
        ])
        sizes.append(x["Product Size"])

    tm_spread = max(tms)-min(tms)
    size_spread = max(sizes)-min(sizes)

    # -----------------------------------
    # Final score
    # -----------------------------------

    penalty = (
        np.mean(pp)*2.0 +
        np.mean(pq)*1.5 +
        np.mean(qq)*1.2 +
        tm_spread*1.5 +
        size_spread*0.05
    )

    score = 100 - penalty

    return {
        "Final Score": round(score,2),
        "Primer-Primer Risk": round(np.mean(pp),2),
        "Primer-Probe Risk": round(np.mean(pq),2),
        "Probe-Probe Risk": round(np.mean(qq),2),
        "Tm Spread": round(tm_spread,2),
        "Size Spread": round(size_spread,2)
    }
```

```python
# ==========================================================
# STEP 4: GENERATE ALL COMBINATIONS
# ==========================================================

records = df.to_dict("records")
combos = list(itertools.combinations(records, PLEX))

results = []

for combo in combos:

    names = " + ".join([x["Name"] for x in combo])

    sc = score_combo(combo)

    row = {"Combination": names}
    row.update(sc)

    results.append(row)

# ==========================================================
# STEP 5: SHOW RESULTS
# ==========================================================

res = pd.DataFrame(results)
res = res.sort_values("Final Score", ascending=False).head(TOP_RESULTS)

display(res)

# ==========================================================
# STEP 6: SAVE EXCEL
# ==========================================================

res.to_excel("Multiplex_Ranked_Results.xlsx", index=False)

print("Saved: Multiplex_Ranked_Results.xlsx")
```
